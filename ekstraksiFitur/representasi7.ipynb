```python
def extract_audio_features(audio_path):
    """Ekstrak fitur dari satu file audio."""
    y, sr = librosa.load(audio_path, sr=None)

    # Zero Crossing Rate
    zcr = np.mean(librosa.feature.zero_crossing_rate(y))

    # Mean Amplitude
    mean_amp = np.mean(np.abs(y))

    # Variance
    variance = np.var(y)

    # Skewness
    skewness = scipy.stats.skew(y)

    # Kurtosis
    kurtosis = scipy.stats.kurtosis(y)

    # Autocorrelation (mengambil lag pertama)
    autocorr = np.correlate(y, y, mode='full')[len(y) - 1] / len(y)

    # Root Mean Square (RMS)
    rms = np.mean(librosa.feature.rms(y=y))
    

    return {
        "File": os.path.basename(audio_path),
        "Zero Crossing Rate": zcr,
        "Mean Amplitude": mean_amp,
        "Variance": variance,
        "Skewness": skewness,
        "Kurtosis": kurtosis,
        "Autocorrelation": autocorr,
        "RMS": rms
    }

def process_audio_folder(folder_path):
    """Memproses semua file audio dalam folder."""
    features_list = []
    for file_name in os.listdir(folder_path):
        if file_name.endswith(".wav"):  # Pastikan hanya membaca file .wav
            file_path = os.path.join(folder_path, file_name)
            features = extract_audio_features(file_path)
            features_list.append(features)

    return features_list

# Path folder audio
folder_path = "../Dataset Audio Angka/7"

# Proses semua audio dalam folder
all_features = process_audio_folder(folder_path)

# Cetak hasilnya
for features in all_features:
    print(features)
```

```python
def calculate_min_max(features_list):
    min_max_values = {}
    keys = features_list[0].keys() if features_list else []
    for key in keys:
        if key != "File":  # Menghindari kolom nama file
            values = [features[key] for features in features_list]
            min_max_values[key] = {"min": min(values), "max": max(values)}
    return min_max_values

# Path folder audio
folder_path = "../Dataset Audio Angka/7"
all_features = process_audio_folder(folder_path)
min_max_values = calculate_min_max(all_features)

# Cetak hasilnya
for feature, values in min_max_values.items():
    print(f"{feature}: Min = {values['min']}, Max = {values['max']}")
```
